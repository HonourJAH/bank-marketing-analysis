```python
# Q1 - Overall result and calling effort

# Question: how many customers subscribed to a term deposit, and how much calling effort did it take?

# Baseline for every later question: the overall yes rate. Rows are records (the data has no customer ID), so "customers" here means records.
```

```python
import pandas as pd

CLEAN_PATH = "../../data/bank_marketing_clean.parquet"
clean_df = pd.read_parquet(CLEAN_PATH)

clean_df.shape
```

```python
# Q1, step 1: overall result and calling effort
contacts = clean_df["contacts_this_campaign"]
subscribed = clean_df["subscribed_term_deposit"] == "yes"

effort_by_result = (
    clean_df.groupby("subscribed_term_deposit")["contacts_this_campaign"]
    .agg(customers="count", total_contacts="sum",
         average_contacts="mean", median_contacts="median")
)
effort_by_result.loc["all"] = [len(clean_df), contacts.sum(), contacts.mean(), contacts.median()]
effort_by_result = effort_by_result.astype({"customers": int, "total_contacts": int})

effort_by_result["share_of_customers_pct"] = (
    effort_by_result["customers"] / len(clean_df) * 100
).round(2)
effort_by_result["share_of_contacts_pct"] = (
    effort_by_result["total_contacts"] / contacts.sum() * 100
).round(2)
effort_by_result["average_contacts"] = effort_by_result["average_contacts"].round(2)

print(effort_by_result)
print()
print(f"Contacts made per subscription: {contacts.sum() / subscribed.sum():.1f}")
```
